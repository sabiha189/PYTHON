Arithmatic Operator

```python
x= 10; y=3
sum= x+y #summation 
sub= x-y #subtraction
mul= x*y #multiplication
div=round(x/y, 2) #division

rem = x%y #remainder
exp = x**y #exponential

print (sum, sub, mul, div, rem, exp)
```

 Logical Operator

```python
x= 10; y=3
greater_than = x>y #greater than
greater_than_equal = x>=y #greater than equal
print (greater_than, greater_than_equal)


x= 10; y=3
less_than = x<y #less than
less_than_equal = x<=y #less than equal
print (less_than, less_than_equal)


equal = x==y #equal
not_equal = x!=y #not equal
print (equal, not_equal)
```

Comparison Operator

```python
x=10; y=5; z=3

result = (x>y) and (y>z) #logical AND
print(result)



x=10; y=5; z=3

result = (x>y) or (y>z) #logical OR
print(result)
```

Conditionals and Statement in Python

if else, nested if else

```python
raining = False
muddy = True


if raining == True:
    print("Take an umbrella")
    if muddy == True:
        print("Wear boots")
else:
    print("Don't take an umbrella")
```
